```python
df[['Attrition','Department']]
```

```python
sns.countplot(x='Attrition', hue='JobRole', data=df)
plt.title('Attrition by JobRole')
plt.ylabel('Count')
plt.xlabel('Attrition')
plt.show()
```

```python
sns.countplot(x='Attrition', hue='OverTime', data=df)
plt.title('Attrition by OverTime')
plt.ylabel('Count')
plt.xlabel('Attrition')
plt.show()
```
